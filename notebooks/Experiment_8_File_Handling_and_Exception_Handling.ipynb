# 🎯 Aim

To understand file handling operations and exception handling in Python by creating, reading, processing files, and managing runtime errors using built-in and custom exceptions.

# 📘 Theory

File handling enables Python programs to store and retrieve data from external files. Python provides functions such as `open()`, `read()`, `write()`, and `close()` to perform file operations. The `with` statement is commonly used because it automatically closes the file after use.

Exception handling allows a program to continue execution even when runtime errors occur. Python uses `try`, `except`, `else`, and `finally` blocks to catch and handle exceptions gracefully. Custom exceptions can also be created by inheriting from the `Exception` class.

## Question 1
### Problem Statement

Create a file named name.txt containing names (one per line), then:

- Count the total number of names.
- Count names starting with a vowel.
- Find the longest name.

```python
n = int(input("Enter number of names: "))

with open("name.txt", "w", encoding="utf-8") as file:
    for _ in range(n):
        file.write(input("Enter name: ") + "\n")

with open("name.txt", "r", encoding="utf-8") as file:
    names = [line.strip() for line in file if line.strip()]

vowels = "AEIOUaeiou"

print("Total Names:", len(names))
print("Names Starting with Vowel:", sum(1 for name in names if name[0] in vowels))
print("Longest Name:", max(names, key=len))
```

### Explanation

The program writes names to a file, reads them back, and performs simple analysis using list operations.

## Question 2
### Problem Statement

Store integers in a file and determine:

- Maximum number
- Average of all numbers
- Count of numbers greater than 100

```python
numbers = list(map(int, input("Enter integers separated by spaces: ").split()))

with open("numbers.txt", "w", encoding="utf-8") as file:
    for num in numbers:
        file.write(str(num) + "\n")

with open("numbers.txt", "r", encoding="utf-8") as file:
    values = [int(line.strip()) for line in file]

print("Maximum:", max(values))
print("Average:", sum(values) / len(values))
print("Greater than 100:", sum(1 for num in values if num > 100))
```

### Explanation

Numbers are written to a file, read back as integers, and analysed using built-in functions.

## Question 3
### Problem Statement

Store city details in a file and:

- Display all cities.
- Display cities with population greater than 10 lakhs.
- Display the total area of all cities.

```python
city_records = [
    ("Dehradun", 5.78, 308.20),
    ("Delhi", 190.00, 1484.00),
    ("Mumbai", 124.00, 603.40),
    ("Pune", 74.00, 331.26),
    ("Jaipur", 41.00, 484.60),
]

with open("city.txt", "w", encoding="utf-8") as file:
    for city, population, area in city_records:
        file.write(f"{city} {population} {area}")

parsed_cities = []
with open("city.txt", "r", encoding="utf-8") as file:
    for line in file:
        parts = line.split()
        if len(parts) == 3:
            city = parts[0]
            population = float(parts[1])
            area = float(parts[2])
            parsed_cities.append((city, population, area))

print("All city details:")
for city, population, area in parsed_cities:
    print(f"{city}: population={population} lakh, area={area} sq km")

high_population_cities = [city for city, population, _ in parsed_cities if population > 10]
total_area = sum(area for _, _, area in parsed_cities)

print("Cities with population > 10 lakhs:", high_population_cities)
print("Sum of all city areas:", round(total_area, 2), "sq km")
```

### Explanation 

Each line is split into fields, converted to correct data types, and then filtered/aggregated.

## Question 4
### Problem Statement
Input N test cases of a b, perform integer division a/b, and handle exceptions for ZeroDivisionError and ValueError.

```python
n = int(input("Enter number of test cases: "))

for _ in range(n):
    try:
        a, b = input("Enter a and b: ").split()
        print(int(a) // int(b))
    except (ZeroDivisionError, ValueError) as e:
        print("Error Code:", e)
```

### Explanation

Each division is placed inside a try-except block so that an error in one test case does not stop the execution of the remaining test cases.

## Question 5
### Problem Statement

Create suitable custom exceptions for a file handling program.

Keep the instructor's custom exceptions:

- EmptyFileError
- InvalidRecordError

```python
class EmptyFileError(Exception):
    pass


class InvalidRecordError(Exception):
    pass


def read_student_marks(file_name):
    try:
        with open(file_name, "r", encoding="utf-8") as file:
            lines = [line.strip() for line in file if line.strip()]
    except FileNotFoundError as error:
        raise FileNotFoundError(f"{file_name} not found.") from error

    if not lines:
        raise EmptyFileError(f"{file_name} is empty.")

    marks_data = {}
    for line in lines:
        parts = line.split(",")
        if len(parts) != 2:
            raise InvalidRecordError(f"Invalid line format: '{line}'")

        name = parts[0].strip()
        try:
            marks = float(parts[1].strip())
        except ValueError as error:
            raise InvalidRecordError(f"Marks must be numeric in line: '{line}'") from error

        marks_data[name] = marks

    return marks_data


with open("students_valid.txt", "w", encoding="utf-8") as file:
    file.write("Aditi,88" 
    "Rohan,76")

with open("students_invalid.txt", "w", encoding="utf-8") as file:
    file.write("Aditi-88\nRohan,abc")

for candidate in ["missing.txt", "students_invalid.txt", "students_valid.txt"]:
    try:
        data = read_student_marks(candidate)
        print(f"{candidate} -> {data}")
    except (FileNotFoundError, EmptyFileError, InvalidRecordError) as error:
        print(f"{candidate} -> {error}")
```

# 🏏 Cricket Analytics Extension

Store cricket player statistics in a file, read the data, calculate the total runs, and identify the highest scorer while handling possible file-related errors.

```python
try:
    with open("players.txt", "w", encoding="utf-8") as file:
        file.write("Virat Kohli,82\n")
        file.write("Shubman Gill,64\n")
        file.write("KL Rahul,41\n")

    players = {}

    with open("players.txt", "r", encoding="utf-8") as file:
        for line in file:
            name, runs = line.strip().split(",")
            players[name] = int(runs)

    print("🏏 Cricket Analytics")
    print("--------------------")

    for player, runs in players.items():
        print(f"{player}: {runs} runs")

    highest = max(players, key=players.get)

    print("\nHighest Scorer:", highest)
    print("Runs:", players[highest])
    print("Team Total:", sum(players.values()))

except FileNotFoundError:
    print("Player data file not found.")
```
