# 📘 Theory

A **string** is a sequence of characters enclosed in single, double, or triple quotes. Python provides numerous built-in methods for string manipulation, making it easy to search, modify, split, and analyse text.

A **set** is an unordered collection of unique elements. Sets are commonly used to remove duplicate values and perform mathematical operations such as union, intersection, difference, and symmetric difference.

Strings and sets are widely used in text processing, data analysis, search algorithms, and duplicate detection.

# Question 1
## Problem Statement

Count and display the number of capital letters in a given string.

```python
text = input("Enter a string: ")

count_caps = sum(1 for ch in text if ch.isupper())

print("Capital letters:", count_caps)
```

## Explanation

This program checks every character using isupper() and counts the uppercase letters.

## Question 2
### Problem Statement

Count the total number of vowels in a given string.

```python
text = input("Enter a string: ")

vowels = "aeiouAEIOU"

count = sum(1 for ch in text if ch in vowels)

print("Total vowels:", count)
```

## Question 3
### Problem Statement

Input a sentence and print each word on a separate line.

## Question 4
### Problem Statement

Count how many times a substring occurs in a string.

```python
main_string = input("Enter the main string: ")
sub_string = input("Enter the substring: ")

count = 0
start = 0

while True:
    index = main_string.find(sub_string, start)

    if index == -1:
        break

    count += 1
    start = index + 1

print("Occurrences:", count)
```

## Question 5
### Problem Statement

Count the frequency of each alphabet (case-insensitive).

```python
text = input("Enter a string: ")

frequency = {}

for ch in text:
    if ch.isalpha():
        key = ch.upper()
        frequency[key] = frequency.get(key, 0) + 1

for key in sorted(frequency):
    print(f"{key} : {frequency[key]}")
```

## Question 6
### Problem Statement

Count the number of unique words in a sentence using sets.

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

unique_words = set(words)

print("Unique Words:", unique_words)
print("Count:", len(unique_words))
```

## Question 7
### Problem Statement

Create two sets of fruits and perform set operations.

```python
s1 = {"apple", "banana", "mango", "grapes"}
s2 = {"banana", "orange", "kiwi", "mango"}

print("Common Fruits:", s1 & s2)
print("Only in Set 1:", s1 - s2)
print("Total Unique Fruits:", len(s1 | s2))
```

## Question 8
### Problem Statement

Perform different set operations on two sets.

```python
S1 = {"Red", "Yellow", "Orange", "Blue"}
S2 = {"Violet", "Blue", "Purple"}

print("Union:", S1 | S2)
print("Intersection:", S1 & S2)
print("Difference (S1-S2):", S1 - S2)
print("Difference (S2-S1):", S2 - S1)
print("Symmetric Difference:", S1 ^ S2)
```

# 🏏 Cricket Analytics Extension

Use strings and sets to analyse cricket player names and identify unique players from two different teams.

```python
team_a = {"Virat Kohli", "Shubman Gill", "KL Rahul", "Hardik Pandya"}

team_b = {"Shubman Gill", "Rohit Sharma", "Hardik Pandya", "Rishabh Pant"}

print("🏏 Cricket Analytics")
print("----------------------")

print("Common Players:")
print(team_a & team_b)

print("\nPlayers only in Team A:")
print(team_a - team_b)

print("\nTotal Unique Players:")
print(len(team_a | team_b))
```
