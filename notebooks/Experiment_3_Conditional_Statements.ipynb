# 🎯 Aim

To understand and implement Python conditional statements (`if`, `elif`, and `else`) for solving decision-making and logical problems based on different conditions.

# 📘 Theory

Conditional statements allow a program to make decisions based on whether a condition is true or false. Python provides `if`, `elif`, and `else` statements to control the flow of execution.

The conditions are evaluated using comparison and logical operators. Based on the result, different blocks of code are executed.

Conditional statements are widely used in applications such as grading systems, login validation, eligibility checks, banking software, and decision-making systems.

## Question 1

### Problem Statement

Check whether a given number is divisible by both 3 and 5.

```python
num = int(input("Enter a number: "))

if num % 3 == 0 and num % 5 == 0:
    print(num, "is divisible by both 3 and 5")
else:
    print(num, "is not divisible by both 3 and 5")
```

### Explanation:

This program checks whether the entered number is divisible by both 3 and 5 using the logical and operator.

## Question 2

Check whether a number is a multiple of 5.

```python
num = int(input("Enter a number: "))

if num % 5 == 0:
    print(num, "is a multiple of 5")
else:
    print(num, "is not a multiple of 5")
```

## Question 3

Find the greatest among two numbers.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print("Greatest number is", a)
elif b > a:
    print("Greatest number is", b)
else:
    print("Both numbers are equal.")
```

## Question 4

Find the greatest among three numbers.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b and a > c:
    print("Greatest number is", a)
elif b > a and b > c:
    print("Greatest number is", b)
else:
    print("Greatest number is", c)
```

## Question 5

Find the roots of a quadratic equation.

```python
import cmath

a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

discriminant = b**2 - 4*a*c

root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
root2 = (-b - cmath.sqrt(discriminant)) / (2*a)

if discriminant >= 0:
    print("The roots are real.")
else:
    print("The roots are imaginary.")

print("Root 1:", root1)
print("Root 2:", root2)
```

## Question 6

Check whether a year is a leap year.

```python
year = int(input("Enter year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

## Question 7

Display the next date

```python
from datetime import datetime, timedelta

date_input = input("Enter date (DD-MM-YYYY): ")

current_date = datetime.strptime(date_input, "%d-%m-%Y")
next_date = current_date + timedelta(days=1)

print("Next Date:", next_date.strftime("%d-%m-%Y"))
```

## Question 8
Print the grade sheet of a student from marks of five subjects.

- Percentage = (total marks / 500) * 100
- CGPA = Percentage / 10
- Grade is assigned using the provided CGPA ranges.

```python
def get_grade(cgpa):
    if 0.0 <= cgpa <= 3.4:
        return "F"
    if 3.5 <= cgpa <= 5.0:
        return "C+"
    if 5.1 <= cgpa <= 6.0:
        return "B"
    if 6.1 <= cgpa <= 7.0:
        return "B+"
    if 7.1 <= cgpa <= 8.0:
        return "A"
    if 8.1 <= cgpa <= 9.0:
        return "A+"
    if 9.1 <= cgpa <= 10.0:
        return "O (Outstanding)"
    return "Invalid CGPA"

name = "Maanav Singh"
roll_no = "R2142220977"
sap_id = "500108304"
semester = 1
course = "B.Tech. CSE CCVT"

marks = {
    "PDS": 70,
    "Python": 80,
    "Chemistry": 90,
    "English": 60,
    "Physics": 65,
}

total = sum(marks.values())
percentage = (total / (len(marks) * 100)) * 100
cgpa = round(percentage / 10, 1)
grade = get_grade(cgpa)

print("Sample Gradesheet")
print("Name:", name)
print("Roll Number:", roll_no, "SAPID:", sap_id)
print("Sem:", semester, "Course:", course)
print("Subject name: Marks")
for subject, score in marks.items():
    print(f"{subject}: {score}")
print(f"Percentage: {percentage:.2f}%")
print("CGPA:", cgpa)
print("Grade:", grade)
```

# 🏏 Cricket Analytics Extension

Use conditional statements to determine a batter's performance based on the number of runs scored.

```python
player = input("Enter player name: ")
runs = int(input("Enter runs scored: "))

if runs >= 100:
    performance = "Century"
elif runs >= 50:
    performance = "Half-Century"
elif runs >= 30:
    performance = "Good Innings"
else:
    performance = "Needs Improvement"

print("\n🏏 Cricket Analytics")
print("--------------------")
print("Player:", player)
print("Runs:", runs)
print("Performance:", performance)
```
