# 🎯 Aim

To understand and implement looping constructs in Python using `for` and `while` loops for solving repetitive computational problems and pattern-based programming tasks.

# 📘 Theory

Loops are used to execute a block of code repeatedly until a specified condition is met. Python mainly provides two looping constructs:

- **for loop:** Used to iterate over sequences such as lists, strings, tuples, and ranges.
- **while loop:** Executes a block of code repeatedly as long as a given condition remains true.

Loops reduce code repetition, improve efficiency, and are widely used in data processing, searching, counting, pattern generation, and algorithm implementation.

## Question 1
### Problem Statement

Find the factorial of a given number using a loop.

```python
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial *= i

print(f"Factorial of {num} is {factorial}")
```

### Explanation

The program multiplies all numbers from 1 to the entered number using a for loop to calculate the factorial.

## Question 2
### Problem Statement

Check whether a given number is an Armstrong number.

```python
num = 153
num_str = str(num)
power = len(num_str)
sum_of_powers = 0

for digit in num_str:
    sum_of_powers += int(digit) ** power

if sum_of_powers == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
```

## Question 3
### Problem Statement

Print the Fibonacci series up to a given number of terms.

```python
terms = 10
a, b = 0, 1

print("Fibonacci series:")
for _ in range(terms):
    print(a, end=" ")
    a, b = b, a + b
print()
```

## Question 4
### Problem Statement

Check whether a given number is prime or not.

```python
num = 29

if num < 2:
    print(num, "is not prime")
else:
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    print(num, "is prime" if is_prime else "is not prime")
```

## Question 5
### Problem Statement

Check whether a given number is a palindrome.

```python
num = 12321

if str(num) == str(num)[::-1]:
    print(num, "is a palindrome")
else:
    print(num, "is not a palindrome")
```

## Question 6
### Problem Statement

Find the sum of digits of a given number.

```python
num = 9876
sum_digits = 0

for digit in str(num):
    sum_digits += int(digit)

print("Sum of digits:", sum_digits)
```

## Question 7
### Problem Statement

Display all numbers between 1 and 100 that are divisible by 5 or 7.

```python
numbers = [n for n in range(1, 101) if n % 5 == 0 or n % 7 == 0]
print("Numbers:", numbers)
print("Count:", len(numbers))
```

## Question 9
### Problem Statement

Print all prime numbers between 1 and 100.

```python
primes = []

for num in range(2, 101):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(num)

print("Prime numbers from 1 to 100:")
print(primes)
```

# 🏏 Cricket Analytics Extension

Use loops to display the runs scored by a cricket player in multiple matches and calculate the total runs.

```python
player = input("Enter player name: ")

matches = int(input("Enter number of matches: "))

total_runs = 0

for i in range(1, matches + 1):
    runs = int(input(f"Runs scored in Match {i}: "))
    total_runs += runs

print("\n🏏 Cricket Analytics")
print("--------------------")
print("Player:", player)
print("Matches:", matches)
print("Total Runs:", total_runs)
print("Average:", total_runs / matches)
```
