# 📘 Theory

Python provides powerful built-in data structures for organizing and managing data efficiently.

- **Lists** are ordered, mutable collections that can store duplicate elements.
- **Tuples** are ordered but immutable collections, making them suitable for fixed data.
- **Dictionaries** store data as key-value pairs, allowing fast retrieval and efficient mapping.

These data structures are widely used in data analysis, software development, and real-world applications such as maintaining records, processing datasets, and generating reports.

## 💻 Experiment Implementation

## Question 1
### Problem Statement

Scan n values in the range 0–3 and print the number of times each value occurs.

```python
n = int(input("Enter number of values: "))

values = []

for i in range(n):
    value = int(input(f"Enter value {i+1} (0-3): "))
    values.append(value)

counts = {0: 0, 1: 0, 2: 0, 3: 0}

for value in values:
    if value in counts:
        counts[value] += 1

for key in counts:
    print(f"{key} occurred {counts[key]} time(s)")
```

### Explanation

The program stores user-entered values in a list and uses a dictionary to count the occurrences of each number from 0 to 3.

## Question 2
### Problem Statement

Create a tuple of numeric values and calculate its average.

```python
numbers = tuple(map(int, input("Enter numbers separated by spaces: ").split()))

average = sum(numbers) / len(numbers)

print("Tuple:", numbers)
print("Average:", average)
```

### Explanation

A tuple is created from the user input, and the average is calculated using sum() and len().

## Question 3
### Problem Statement

Input students' scores and print the runner-up score.

```python
scores = list(map(int, input("Enter scores separated by spaces: ").split()))

unique_scores = sorted(set(scores), reverse=True)

if len(unique_scores) > 1:
    print("Runner-up score:", unique_scores[1])
else:
    print("Runner-up score not available.")
```
