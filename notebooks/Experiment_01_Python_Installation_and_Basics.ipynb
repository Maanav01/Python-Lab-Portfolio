# 🎯 Aim

To install Python successfully, understand the Python development environment, and learn the fundamentals of Python programming, including printing, variables, data types, string operations, and formatted output.

# 📘 Theory

Python is a high-level, interpreted, and object-oriented programming language known for its simple syntax and readability. It is widely used in web development, automation, data science, artificial intelligence, cloud computing, and software development.

Python programs can be executed in two modes:

- **Interactive Mode:** Executes one statement at a time and immediately displays the output. It is useful for testing small pieces of code.
- **Script Mode:** Executes an entire Python program stored in a `.py` file. It is suitable for writing complete applications.

This experiment introduces the basic building blocks of Python, such as printing output, creating variables, working with different data types, and performing simple string operations.
