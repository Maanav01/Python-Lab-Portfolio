## Question 1
### Problem Statement

Create a Student class with:

- Name
- SAP ID
- Marks in Physics, Chemistry and Mathematics

Create three student objects and display:

- Student Details
- Total Marks
- Percentage

```python
class Student:
    def __init__(self, name, sap_id, marks):
        self.name = name
        self.sap_id = sap_id
        self.marks = marks

    def total(self):
        return sum(self.marks)

    def percentage(self):
        return self.total() / len(self.marks)

    def display(self):
        print("\nStudent Details")
        print("Name:", self.name)
        print("SAP ID:", self.sap_id)
        print("Marks:", self.marks)
        print("Total:", self.total())
        print("Percentage:", f"{self.percentage():.2f}%")

students = []

for i in range(3):
    print(f"\nEnter details for Student {i+1}")
    name = input("Name: ")
    sap = input("SAP ID: ")
    marks = list(map(int, input("Enter Physics Chemistry Maths marks: ").split()))
    students.append(Student(name, sap, marks))

for student in students:
    student.display()
```

### Explanation

The Student class stores student information and provides methods to calculate the total marks and percentage.

## Question 2
### Problem Statement

Create a class with a constructor and implement:

- display()
- calculate_percentage()
- find_grade()

```python
class StudentRecord:
    def __init__(self, name, sap_id, marks):
        self.name = name
        self.sap_id = sap_id
        self.marks = marks

    def calculate_percentage(self):
        return sum(self.marks) / len(self.marks)

    def find_grade(self):
        percentage = self.calculate_percentage()

        if percentage >= 90:
            return "A+"
        elif percentage >= 75:
            return "A"
        elif percentage >= 60:
            return "B"
        elif percentage >= 50:
            return "C"
        return "D"

    def display(self):
        print("\nName:", self.name)
        print("SAP ID:", self.sap_id)
        print("Percentage:", f"{self.calculate_percentage():.2f}%")
        print("Grade:", self.find_grade())

name = input("Enter name: ")
sap = input("Enter SAP ID: ")
marks = list(map(int, input("Enter marks: ").split()))

student = StudentRecord(name, sap, marks)
student.display()
```

### Explanation

The constructor initializes the object, while separate methods calculate the percentage and determine the grade.

## Question 3
### Problem Statement

Implement different types of inheritance:

- Single Inheritance
- Multilevel Inheritance
- Multiple Inheritance
- Hierarchical Inheritance

```python
# Single inheritance
class Person:
    def introduce(self):
        print("I am a person.")


class Teacher(Person):
    def subject(self):
        print("I teach Python.")


# Multilevel inheritance
class Animal:
    def breathe(self):
        print("Breathing...")


class Mammal(Animal):
    def feed_milk(self):
        print("Feeding milk...")


class Dog(Mammal):
    def bark(self):
        print("Woof!")


# Multiple inheritance
class Writer:
    def write(self):
        print("Writing an article.")


class Speaker:
    def speak(self):
        print("Speaking at an event.")


class AuthorSpeaker(Writer, Speaker):
    pass


print("Single Inheritance:")
t = Teacher()
t.introduce()
t.subject()

print("Multilevel Inheritance:")
d = Dog()
d.breathe()
d.feed_milk()
d.bark()

print("Multiple Inheritance:")
as_role = AuthorSpeaker()
as_role.write()
as_role.speak()
```

### Explanation 
Inheritance helps us reuse and extend behavior from parent classes. Below examples show single, multilevel, and multiple inheritance.

## Question 4
### Problem Statement

Create a class demonstrating Method Overriding.

```python
class Shape:
    def describe(self):
        print("This is a generic shape.")

class Circle(Shape):
    def describe(self):
        print("This is a Circle.")

shape = Shape()
circle = Circle()

shape.describe()
circle.describe()
```

### Explanation

The child class overrides the describe() method inherited from the parent class, providing its own implementation.

## Question 5
### Problem Statement

Demonstrate Operator Overloading by adding two Point objects.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print(f"({self.x}, {self.y})")

p1 = Point(2, 5)
p2 = Point(4, 7)

result = p1 + p2

print("Result:")
result.display()
```

### Explanation

The __add__() special method overloads the + operator, allowing two Point objects to be added together.

# 🏏 Cricket Analytics Extension

Use object-oriented programming to represent cricket players as objects, calculate their strike rates, and identify the highest scorer.

```python
class Player:
    def __init__(self, name, runs, balls):
        self.name = name
        self.runs = runs
        self.balls = balls

    def strike_rate(self):
        return (self.runs / self.balls) * 100

players = [
    Player("Virat Kohli", 82, 53),
    Player("Chris Gayle", 84, 48),
    Player("AB Devilliers", 90, 37)
]

print("🏏 Cricket Analytics")
print("--------------------")

highest = players[0]

for player in players:
    print(f"{player.name}: {player.strike_rate():.2f}")

    if player.runs > highest.runs:
        highest = player

print(f"\nHighest Scorer: {highest.name} ({highest.runs} runs)")
```
