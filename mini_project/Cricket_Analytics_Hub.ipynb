## Introduction

Cricket generates a huge amount of statistical data during every match. Coaches, analysts and fans rely on player and team statistics to understand performance and make informed decisions.

The Cricket Analytics Hub is a Python-based data analysis project that demonstrates how cricket datasets can be analysed using Pandas, NumPy and Matplotlib. The project focuses on extracting meaningful insights from player statistics and presenting them through visualizations.

## Problem Statement

The aim of this project is to analyse cricket datasets, calculate useful statistics and generate graphical visualizations that help users understand player performance efficiently.

### 📦 Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (10,6)

```

### 📥 Load Dataset

```python
batters = pd.read_csv("../datasets/batter_player_stats.csv")

batters.head()
```

### Dataset Information

```python
batters.info()
```

### Statistical Summary

```python
batters.describe()
```

### Missing Values

```python
batters.isnull().sum()
```

### Data Cleaning

```python
batters.fillna(0, inplace=True)

batters.head()
```

### Number of Players

```python
print("Total Players:",len(batters))
```

### Top 10 Run Scorers

```python
top_players = batters.sort_values(
    by="total_runs",
    ascending=False
).head(10)

top_players[["player_name","team","total_runs"]]
```

### Highest Run Scorer

```python
highest = batters.loc[
    batters["total_runs"].idxmax()
]

highest
```

### Average Runs

```python
print("Average Runs :", batters["total_runs"].mean())
```

### Highest Strike Rate

```python
batters.sort_values(
    by="strike_rate",
    ascending=False
)[["player_name","strike_rate"]].head()
```

### Players with 500+ Runs

```python
batters[
    batters["total_runs"] >= 5000
][["player_name","team","total_runs"]]
```

### Visualization 1
Top 10 Run Scorers

```python
plt.figure(figsize=(12,6))

plt.bar(
    top_players["player_name"],
    top_players["total_runs"]
)

plt.xticks(rotation=45)

plt.title("Top 10 Run Scorers")

plt.xlabel("Player")

plt.ylabel("Runs")

plt.show()
```

### Visualization 2
Strike Rate

```python
top_sr = batters.sort_values(
    by="strike_rate",
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

plt.bar(
    top_sr["player_name"],
    top_sr["strike_rate"]
)

plt.xticks(rotation=45)

plt.title("Top Strike Rate")

plt.xlabel("Player")

plt.ylabel("Strike Rate")

plt.show()
```

### Visualization 3
Runs Distribution

```python
plt.hist(
    batters["total_runs"],
    bins=15
)

plt.title("Distribution of Runs")

plt.xlabel("Runs")

plt.ylabel("Players")

plt.show()
```

### Visualization 4
Pie Chart

```python
top5 = top_players.head(5)

plt.pie(
    top5["total_runs"],
    labels=top5["player_name"],
    autopct="%1.1f%%"
)

plt.title("Top 5 Run Contributors")

plt.show()
```

### Visualization 5
Scatter Plot

```python
plt.scatter(
    batters["total_runs"],
    batters["strike_rate"]
)

plt.xlabel("Runs")

plt.ylabel("Strike Rate")

plt.title("Runs vs Strike Rate")

plt.show()
```

## Insights

- The highest scorer consistently contributes significantly to the team's total runs.
- Players with higher strike rates tend to score more quickly.
- Most players fall within the medium-run range.
- A small number of players contribute a large percentage of the total runs.
